```python
def scenario_name(use_label_embeddings,freeze_emb,random_word_embedding,lr):
    scenario = ''
    if use_label_embeddings:
        scenario += '_use_label_emb'
    if freeze_emb:
        scenario += '_freeze_label_emb'
    if random_word_embedding:
        scenario += '_randomized_word_emb'
    scenario += '_'
    scenario+=str(lr)
    return scenario
```
