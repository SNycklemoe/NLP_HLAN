```python
import numpy as np
import matplotlib.pyplot as plt

# Load the .npy files
files = ["../Data/Training_results/train_loss_100.npy", "../Data/Training_results/dev_micro_f1_100.npy",
        "../Data/Training_results/dev_macro_f1_100.npy","../Data/Training_results/dev_micro_auroc_100.npy",
        "../Data/Training_results/dev_macro_auroc_100.npy"]
file_names = ['Training Loss', 'Dev Micro F1', 'Dev Macro F1','Dev Micro AUROC','Dev Macro AUROC']
data = [np.load(file) for file in files]

# Create a figure with 5 subplots in a row
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Plot each list on its respective subplot
for i, ax in enumerate(axes):
    ax.plot(data[i])
    ax.set_title(file_names[i])

plt.tight_layout()
plt.show()
```

```python
def scenario_name(use_label_embeddings,freeze_emb,random_word_embedding,lr):
    scenario = ''
    if use_label_embeddings:
        scenario += '_use_label_emb'
    if freeze_emb:
        scenario += '_freeze_label_emb'
    if random_word_embedding:
        scenario += '_randomized_word_emb'
    scenario += '_'
    scenario+=str(lr)
    return scenario
```

```python
use_label_embeddings = True
freeze_emb = True
random_word_embedding = True
num_epochs = 250

lr = 0.0001
scenario = scenario_name(use_label_embeddings,freeze_emb,random_word_embedding,lr = lr)


files = ["../Data/Training_results/train_loss_"+str(num_epochs)+scenario+".npy","../Data/Training_results/dev_micro_f1_"+str(num_epochs)+scenario+".npy",
        "../Data/Training_results/dev_macro_f1_"+str(num_epochs)+scenario+".npy","../Data/Training_results/dev_micro_auroc_"+str(num_epochs)+scenario+".npy",
        "../Data/Training_results/dev_macro_auroc_"+str(num_epochs)+scenario+".npy"]
scenario = 'Label/freeze/random'
file_names = ['Training Loss', 'Dev Micro F1', 'Dev Macro F1','Dev Micro AUROC','Dev Macro AUROC']
data = [np.load(file) for file in files]
# Create a figure with 5 subplots in a row
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for i, ax in enumerate(axes):
    print(np.argmax(data[i]))
    ax.plot(data[i], label=scenario)  # Use the same label for all plots
    ax.set_title(file_names[i])
    ax.legend(loc='best')  # Ensure legend is called for each subplot

plt.tight_layout()
plt.show()

# /Users/leojin/Desktop/BDS Course/CS 769/NLP_HLAN/Data/Training_results/train_loss_200_use_label_emb_freeze_label_emb_randomized_word_emb.npy
```
